## map
iterable객체의 각요소에 대해 지정한 함수를 적용하고, 그 결과를 map객체로 반환
```
map(function, iterable) -> map
```




````python
### map 함수: 첫 번째 인자는 함수, 두 번째 인자는 순회할 수 있는 자료형
## -> 순회할 수 있는 자료형에 대해 순차적으로 첫 번째 인자에 해당하는 함수를 실행함.
# 최종적으로 확인을 위해 list로 바꾼 후 확인하여야 함.

nums: list = [1,2,3,4,5]

result = map(lambda x:x*2, nums)
print(result, type(result)) # map 주소, map 타입
print(list(result))
print(set(result))

# map(): 기존 list의 요소를 하나씩 꺼내 함수를 호출,
#        함수 결과들을 모아 새로운 list 생성
````

````python
# 문자열 리스트 -> 문자열 길이 리스트
````

````python
products: list[str] = ["아이스 아메리카노", "수박주스", "말차라떼", "식혜", "딸기 스무디"]

products_len : list[int] = map(lambda x: len(x), products)
print(list(products_len))
````

````python
# 여러 순회 가능 객체(list) 통합
mid_term = [90, 55, 70]
final_term = [100, 40, 75]

avg_scores = map(lambda x,y:(x+y)/2, mid_term, final_term)
print(list(avg_scores))
````

## filter

주어진 iterable객체에 대해서 참/거짓 평가후 참인 요소들만 결과값으로 반환

```
filter(function, iterable) -> filter
```

````python
# 순회 가능한 자료형에 대해 순차적으로 함수에 대입하였을 때 그 값이 참인 경우에만 값을 저장하며,
# 그 외의 값은 제거한다. 말 그대로 필터링이다.

nums = list(range(1,11))
print(nums)

evens = filter(lambda n: n % 2 == 0, nums)
odds = filter(lambda n: n % 2 != 0, nums)
threes = filter(lambda n: n % 3 == 0, nums)

print((evens, type(evens)))
print("list(evens):", list(evens))
print("list(odds):", list(odds))
print("list(threes):", list(threes))
````

````python
# 이름 목록에서 성이 "홍"씨인 사람만 찾아서 list로 반환
names: list[str] = ["홍길동", "고길동", "홍진경","최홍만", "홍명보"]

# str.startswith(str2) -> bool
# - 문자열의 시작이 str2와 같다면 true
hong_family =  list(filter(lambda name: name.startswith("홍"), names))
print(hong_family)
````

````python
# dict 검색(dict 필터링)
# 학생 목록에서 점수가 90점 이상인 학생만 필터링
students = [
    { 'name' : '이순신', 'score' : 80},
    { 'name' : '윤봉길', 'score' : 75},
    { 'name' : '논개', 'score' : 99},
    { 'name' : '강감찬', 'score' : 91},
]

# filter() 함수
students1 = list(filter(lambda s: s["score"] >= 90,students))
print(students1)

# list comprehension
students2 = [s for s in students if s["score"] >= 90]
print(students2)
````

## reduce
iterable객체의 요소에 대한 연산결과를 단일값을 반환.
합계, 최대/최소값 등의 연산에 적합

```
reduce(function, iterable, initial_value) -> Any
```
함수는 3개의 인자를 받는다.
1. 누적 연산값
2. 현재 요소
3. 초기값 (선택)


````python
from functools import reduce
# 외부 functools에서 import reduce

# 정수형 list 내 요소들의 합계
nums: list[int] = [11,22,33,44,55,66,77,88,99]

# 총합 저장
# total = reduce(콜백함수(누적용변수, 현재값), 순회객체, 누적변수초기값) -- 누적변수초기값: 처음 변수에 저장될 초기값 (생략하면 1행 값 들어감)
## -> 콜백함수의 return 값이 자동으로 누적용 변수에 대입된다.
total = reduce(lambda sum, curr: sum + curr, nums, 0)
print(total)
````

````python
# 최대, 최솟값 찾기
nums: list[int] = [32,123,-48,5,2,6,-489,894,5,52,31]

# def func(max, curr):

#최대값
max_num: int = reduce(lambda max, curr: max if max > curr else curr ,nums, 0)
print(max_num)

#최소값
min_num: int = reduce(lambda min, curr: min if min < curr else curr, nums, 0)
print(min_num)

#절대값이 가장 큰 수
def abs_func(max, curr):
    print(f"max: {max}, curr: {curr}")
    return max if abs(max) > abs (curr) else curr

abs_max = reduce(abs_func, nums, 0)
print(abs_max)
````
