# dict(Dictionary, 사전 자료형)
- `dict`는 하나의 요소를 Key : Value 형식으로 갖는 자료형이다.
- 사전처럼 단어를 알면 뜻을 알 수 있듯이, Key를 통해 Value를 확인하는 구조
- list 처럼 index로 접근하는 것이 아닌 `Key`로 접근 가능

## 1. dict 생성과 key 규칙

`dict`의 핵심은 `key`이다.

- 하나의 `dict` 안에서 `key`는 중복될 수 없다.
- 같은 `key`를 다시 넣으면 마지막 값으로 덮어써진다.
- `key`는 값을 빠르게 찾기 위한 기준이므로 중간에 값이 바뀌면 안 된다.
- 그래서 `int`, `float`, `str`, `tuple` 같은 불변 자료형은 `key`로 사용할 수 있지만, `list`, `dict`, `set` 같은 가변 자료형은 `key`로 사용할 수 없다.
- Python 3.7 이상에서 `dict`는 삽입 순서를 유지한다.

```python
print("--- dict ---")
dct = {
    "a":10,
    "b":20,
    "c":100
}

print(dct, type(dct))

# key는 불변 타입만 가능
dct2 = {
    10 : "aaa",
    (10,20) : "bbb"
}

print(dct2, type(dct2))

# # key 가변 타입 불가 확인
# dct3 = {
#     [1,2,3] : "CCC"
# }
```

```python
print("--- dict 조회 ---")
print("dct:", dct)

# key를 이용한 조회
print("dct['a']:", dct["a"])
print("dct.get('a'):", dct.get("a"))

# 없는 Key 조회
# print("dct['d']:", dct["d"]) # KeyError
print("dct.get('d'):", dct.get("d")) # get() 방식은 없으면 None을 반환함.

# 없으면 기본값 적용
print("dct.get('d', '값 없음'):", dct.get("d", "값 없음"))
# 없으면 기본값 적용
print("dct.get('c', '값 없음'):", dct.get("c", "값 없음"))
```

## dict 요소 추가, 수정, 삭제
- dict는 mutable 자료형
- dict[없는key] = value로 value가 추가
- dict[key] = value로 value 수정 가능
- dict.update()를 이용해 value 수정 가능 (여러 값 동시 수정도 가능)
- pop() 또는 del을 이용해 요소 제거

```python
dct["e"] = 888
print("dct['e']:", dct["e"])
print(dct)

dct["e"] = 999
print("dct['e']:", dct["e"])
print(dct)

# update()
dct.update(e = 1234)
print("dct['e']:", dct["e"])
print(dct)

# 여러 value 업데이트
# dct.update(e = 528, a = 444, c = 85888)
dct.update( { "a": 20, "b": 30 } )
print("dct['e']:", dct["e"])
print(dct)

# 요소 제거
# 1) del 연산
# del dct["e"]
# 2) pop() 메서드 == 꺼내기
A = dct.pop("b")
print(dct)
print("A:", A)
```

```python
# dict 생성 함수와 주요 메서드
# - dict 생성 방법
# 1) {k:v, k:v, ... }
# 2) dict(key=value, key=value)
dct3 = dict(name='홍길동', age=22)
print(dct3)

# 주요 메서드
keys = dct3.keys() # key만 추출
print(keys, type(keys)) # dict_keys 타입

values = dct3. values() # 값만 추출
print(values, type(values)) # dict_values 타입

items = dct3.items() # key - value 쌍 형태 추출 --> tuple로 묶인 형태
print(items, type(items))
```

# dict 순회와 unpacking
- `dict` 반복문(for) 바로 넣으면 key를 하나씩 반환한다.
- key가 아닌 value 또는 item을 하나씩 반환 받기 위해서는 dict 메서드 활용

```python
dct4 = {
    "홍길동": 20,
    "이순신": 30,
    "신사임당":40,
    "강감찬":70
}

print('-- dict 순회 기본 ---')
for key in dct4:
    print(f"이름: {key} / 나이: {dct4[key]}세")

print("--- dict.keys() 순회 ---")
for key in dct4.keys():
    print(f"이름: {key} / 나이: {dct4[key]}세")

print("--- dict.values() 순회 ---")
for value in dct4.values():
    print(value)

print("--- dict.items() 순회 ---")
for key, value in dct4.items():     # unpacking
    print(key, value)

for item in dct4.items():       # tuple 반환
    print(item)
```

# dict 대입, 얕은 복사, 깊은 복사

- 대입: 같은 객체를 함께 바라보도록 참조를 복사한다.
- 얕은 복사(shallow copy): 새 컨테이너 객체를 만들지만, 내부 중첩 객체는 공유할 수 있다.
- 깊은 복사(deep copy): 내부 중첩 객체까지 재귀적으로 새로 복사한다.
- 중첩 객체가 없는 단순 dict에서는 copy()가 깊은 복사처럼 보일 수 있지만, 정확히는 얕은 복사이다.

```python
sample = {                              # 실제 데이터 저장 위치: 딕셔너리의 모든 내용은 별도의 주소에 저장되며,
    'name' : '기계식 키보드',             # 딕셔너리 변수(sample)에는 해당 주소를 바라볼 수 있는 동일한 주소가 들어있음
    'price' : 30000,
    'origin' : 'KR'
}

print("--- dict 대입: 새 변수에 주소 복사(같은 객체 참조) ---")
other = sample                          # 이 경우 dictionary가 복사되는 것이 아닌 주소가 복사된 것.
print('id(other)', id(other))
print('id(sample)', id(sample))         # 같은 주소를 바라보고 있는 얕은 복사 (shallow copy)

print("before sample['name']: ", sample['name'])
print("before other['name']: ", other['name'])

sample['name'] = "비싼 키보드"

print("after sample['name']: ", sample['name'])
print("after other['name']: ", other['name'])
```

```python
print("--- dict copy(): 얕은 복사 ---")
print("copy() 된 dict의 value 중 immutable 타입의 값은 공유 X")

other2 = sample.copy()          # .copy(): 얕은 복사를 명시적으로 입력 --------> 단순 = 연산자와 달리

print("id(sample)", id(sample))
print("id(other2)", id(other2))
print("sample is other2:", sample is other2)

# sample["price"] = sample["price"] * 2
sample["price"] *= 2            # 우항이 먼저 실행되고 그 반환값을 좌항에 저장

print("sample['price'] 변경 후: ", sample['price'])
print("other2['price'] 변경 후: ", other2['price'])
```

```python
print("--- dict copy() 확인 ---")
print("copy() 된 dict의 value 중 mutable 타입의 값은 공유 X")
sample2 = {
    "name" : "노트북",
    "options" :["32GB RAM", "1TB SSD"]  # list: mutable (변경 가능함)
}

other3 = sample2.copy()

sample2["options"].append("RTX 5090")
# sample2["options"].append(True)
print("sample2:", sample2)
print("other3:", other3)
print(id(sample2), id(other3))
```
