# 예외-오류 Exception-Error

- Exception : 정상적인 흐름으로 복구가능한 미약한 오류. (try-except구문으로 복구)

## 예외처리
**(중요) 발생된 예외/오류 상황에서 try..except구문을 통해 적절히 처리후 정상적인 흐름을 돌려내는 것**

```python
print("시작")

# try: 예외가 발생할 가능성이 있는 코드를 시도하는 구문 ("에라가 발생할 수 있지만 일단 시도")
try:
    result = 10 / 0 # 예외가 발생할 수 있는 코드

    # 예외 발생 원리
    # 1. 예외(오류로서) 감지
    # 2. 해당 예외 상황에 맞는 "예외 클래스"를 찾음 --> 클래스를 가지고는 무엇을 할 수 없음 --> 객체를 생성해야 함
    # 3. 예외 클래스를 이용해서 "예외 객체"를 생성
    # 4. 생성된 예외 객체를 예외 상황이 감지된 코드 위치로 올림<raise> (클래스를 올림 X./ 객체를 올림 O)

    ## 결론: try가 시도 후 문제가 생기면 예외 객체를 생성 후 올린다.

except:
    # except: try에서 올린 예외 처리 구문을 실행하는 구문
    print("0으로 나눌 수 없습니다.")

    # except 구문이 종료되면 다시 정상 흐름(다음 순서)으로 복귀



print("종료")
```

```python
# except 구문을 이용한 예외 객체(try에서 실패했을 때 올렸던) 사용 방법
import random # 난수 생성 모듈

try:
    num = random.randint(0,1) # 0~1 사이 정수형 난수 생성
    print("num:", num)

    # 0 == False, 1 == True
    if num: # 1인 경우
        print("abc" + 123)  # TypeError
    else: # 0인 경우
        print(int("abc"))   # ValueError

# e: try에서 발생한 예외 객체
except TypeError as e:
    print(e)
    print("예외 타입 이름:", type(e).__name__)    # type함수를 사용해서 나온 결과에서 이름만 뽑을 때 __nama__을 사용함
    print("예외 메시지:", str(e))
    print("예외 상세 값:", e.args)
    print(f"[Error] {type(e).__name__}: {str(e)}")


except ValueError as e:
    print(e)
    print("예외 타입 이름:", type(e).__name__)    # type함수를 사용해서 나온 결과에서 이름만 뽑을 때 __nama__을 사용함
    print("예외 메시지:", str(e))
    print("예외 상세 값:", e.args)
    print(f"[Error] {type(e).__name__}: {str(e)}")
```

```python
# 숫자를 입력 받아 입력 받은 숫자와 key가 일치하는 학생을 조회하여 id, name 출력
# 입력 받은 숫자가 students에 없으면 "1~3 사이 숫자만 입력해주세요"를 출력
# + 잘못된 숫자 입력 시 다시 숫자를 입력받게 하기

students: dict[int, dict] = {
    1: {"id": 10, "name": "홍길동"},
    2: {"id": 20, "name": "신사임당"},
    3: {"id": 30, "name": "이순신"},
}

while True:
    try:
        # print("""
        #     10: 홍길동
        #     20: 신사임당
        # """)
        num = int(input("숫자를 입력해주세요."))
        student:dict = students[num]    # 선택된 key 값에 대한 value를 반환하여 student에 저장
        # print(student)
        print("id:", student["id"])
        print("name:", student["name"])
        break   # 정상 입력 시 반복 종료

    except KeyError as e:
        print("1~3 사이 숫자만 입력해주세요")

print("--- 프로그램 종료👍 ---")
```

### pass

- 이 때, 구조만 유지하는 용도 또는 예외를 무시할 때 사용

```python
if True:
    pass    # 코드 내부가 비어있을 경우의 오류 무시 및 구조 유지를 위함

def func():
    pass
```

```python
# 예외 무시
try:
    print(10/0)
except:
    pass

print("종료")
```

### finally
- try 구문에서 예외 발생 여부와 관계 없이 마지막에 무조건 실행하는 구문 (블록)

### else
- except 구문을 모두 지나친 경우 수행되는 구문 / try 구문에서 예외가 발생되지 않았을 때 수행되는 구문

```python
try:
    num = int(input("숫자 입력"))
    result = 100 / num

except ValueError as e:
    print("정수형 숫자만 입력하세요")

except ZeroDivisionError as e:
    print("0으로 나눌 수 없습니다.")

else:   # try 구문에서 예외가 발생하지 않았다면? --> else로 넘어감, 즉 try 다음에 에러 없으면 바로 else로 넘어간다고 보면 됨
    print("계산 결과:", result)

finally:        # try에서 에러가 나든 안 나든 무조건 마지막에 실행됨
    print("예외 발생 여부와 관계 없이 무조건 실행됨")
```

```python
# finally 사용 예시
try:
    f = open("helloworld.txt", "r", encoding="utf-8")
    print(f.read())
    print(10/0) # 예외 발생 상황
except:
    pass

finally:
    f.close()   # 메모리 반환 (필수)
```
