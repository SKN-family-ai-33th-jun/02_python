# lambda 람다

- `lambda`는 함수를 한 줄로 간결하게 선언하는 문법이다.
- 이름이 없는 작은 함수를 만들 때 사용한다.
- 기본 구조는 `lambda 매개변수: 반환값`이다.
- 복잡한 로직은 일반 `def` 함수로 작성하는 편이 읽기 쉽다.
```
변수명 = lambda (매개변수): (함수 실행 내용)
# 함수 실행 내용에 작성된 코드(표현식)이 끝나면 자동으로 반환된다
```

## 1. lambda 기본 문법

아래 두 코드는 같은 의미이다.

```python
def square(x):
    return x ** 2

square = lambda x: x ** 2
```

````python
def square(x: int) -> int:
    return x ** 2

square2 = lambda x : x ** 2
# print( lambda x : x ** 2 )

print(square2(5))
print(square(5))

#####lambda 함수의 구조#####
# lambda, 매개변수, 처리값(return)
# multiple = lambda a,b,c : a*b*c
# multiple은 함수명, lambda는 def, a,b,c는 매개변수, a*b*c는 리턴값
````

## 2. list comprehension 안에서 사용하기

`lambda`는 짧은 계산식을 값처럼 전달하거나 즉시 실행할 때 사용할 수 있다.

````python
def make_squares() -> list[int]:
    return [(lambda x: x**2)(n) for n in range(11)]

print(make_squares())
````

## 3. 여러 매개변수 사용

`lambda`도 일반 함수처럼 하나 이상의 매개변수를 받을 수 있다.

````python
add = (lambda a, b, c : a + b + c)(1,2,3)
print(add)


multiple = (lambda x, y: x * y)(2,9)
print(multiple)
````

## 4. 기본값, `*args` 사용

`lambda`에서도 기본값이 있는 매개변수와 `*args`를 사용할 수 있다. 다만 코드가 길어지면 `def` 함수로 분리하는 것이 좋다.

````python
print("--- lambda 매개변수 기본값 ---")

multiply = lambda a, b=8 : a*b
print(multiply(10))

print("--- lambda *args ---")
adder = lambda * args:sum(args)

print(adder(2,3,4,5,6,8))


a = [5,2,1,4,7,8,9]

print(sum(a))
````

## 5. lambda와 고차 함수

고차 함수는 함수를 인자로 받거나, 함수를 반환하는 함수이다.

`lambda`는 이름 없는 함수를 값처럼 만들어서 다른 함수에 전달할 때 자주 사용한다.
아래 예제에서 `apply_to_numbers()`는 `func`라는 함수를 인자로 받는 고차 함수이다.


````python
##### 고차함수: 매개변수 입력으로 함수를 받는 함수
##### 콜백함수: 매개변수로서 사용되는 그 함수 (즉 나중에 매개변수로 동작함으로 인해 다시 콜백되는 함수)

numbers = [1,2,3,4,5]

# 매개변수 func ==  함수를 전달받을 목적임.
def apply_to_numbers(numbers, func):

    result = []
    # 순서: func라는 함수에 n이라는 인자를 넣어 해당 매개변수를 통한 return값을 append 한다.
    # 즉, 기존 과정에서 값이 함수를 한 번 통과하는 것이 추가되었을 뿐.
    for n in numbers:
        result.append( func(n) )
    return result

numbers1 = apply_to_numbers(numbers, lambda x: x**2)

print(numbers1)


numbers2 = apply_to_numbers(numbers, lambda n: n **3)
print(numbers2)

numbers3 = apply_to_numbers([n for n in range(11)], lambda n: n*21)
print(numbers3)
````

````python
# 함수를 반환하는 고차 함수
def make_multiplier(n: int):
    return lambda number : number * n

triple = make_multiplier(3)
print(triple(5))
````

## 6. lambda와 콜백 함수

콜백 함수는 내가 바로 실행하는 함수가 아니라, 다른 함수에 전달해 두었다가 필요한 시점에 호출되는 함수이다.

`lambda`는 콜백 로직이 짧을 때 사용하기 좋다. 다만 콜백 안의 코드가 길어지면 이름 있는 `def` 함수로 작성하는 편이 읽기 쉽다.


````python
# 매개변수 operation == 콜백함수
# calculator == 고차함수

def calculator(a,b,operation):
    return operation(a,b)

add_result = calculator(999,888, lambda a,b: a+b)
print(add_result)

minus_result = calculator(999,888, lambda a,b: a-b)
print(minus_result)
````
