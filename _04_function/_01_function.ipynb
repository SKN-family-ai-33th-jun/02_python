# Function 함수

- 함수는 반복되거나 절차적인 실행 코드를 하나의 이름으로 묶은 것이다.
- 함수를 선언한 뒤 호출해서 사용한다.
- 호출할 때 값을 전달할 수 있고, 실행 결과를 반환받을 수 있다.
- Python에서 함수는 하나의 값처럼 다룰 수 있는 1급 객체이다.

## 1. 함수 선언과 호출

`def` 키워드로 함수를 선언하고, 함수 이름 뒤에 괄호를 붙여 호출한다.

```python
from IPython.utils.PyColorize import pride_theme


def add(a,b):   # add 함수 선언
    c= a+b      # add 함수 정의
    return c

print("add() 함수 호출:", add(3,4))     # add 함수 호출(call)
print("type(add):", type(add))
print(add) # add: 함수 객체(선언, 정의) 참조 == 함수 모양 그대로를 보게 함 (괄호 없이 기입)

# 값 전달 0, 결과 반환 X
def minus(a: int, b:int) -> None:
    print("minus() 함수 호출 결과:", a-b)

minus(3,4)

# 값 전달 X, 결과 반환 O
def multiply() -> None:
    return 111*111

print(multiply())

# 값 전달 X, 결과 반환 X
def greeting() -> None:
    print("hello hello")

greeting()
```

## 2. 매개변수와 인수

- **매개변수(parameter)**: 함수 선언부에서 값을 받기 위해 만든 변수
- **인수(argument)**: 함수 호출부에서 실제로 전달하는 값

```python
def foo(k, m): # k,m == 매개변수 (parameter)
    print("k:", k)
    print("m:", m)

foo(100, 200) # 100,200 == 인자 (arguments)
```

## 3. return과 함수 객체

함수는 실행을 마치면 호출한 위치로 돌아온다. 이때 `return`으로 값을 하나 돌려줄 수 있다.

또한 Python에서 함수는 변수에 담거나 다른 함수의 인수로 전달할 수 있다. (1급 객체)

### 1급 객체
- 값 처럼 다룰 수 있는 객체
1. 변수에 대입 가능
2. 함수 인자로 전달 가능
3. 함수 반환 값으로 사용 가능
4. 자료 구조에 저장 가능

```python
##### 함수를 매개변수로 사용하는 대표적 예시

# 일반적인 데이터 타입 return
def bar():
    print("bar")
    return 0

# bar 함수를 func 변수에 대입
func = bar  # 주소 복사
print("bar 주소:", id(bar))
print("func 주소:", id(func))
print("bar is func:", bar is func) # 같은지 비교

# 함수를 인자로 전달
def runner(fff):
    print(fff()) # 파라미터에 담긴 함수를 호출

runner(func)
```

## 4. 위치 인수와 키워드 인수

함수를 호출할 때 인수를 전달하는 방식은 크게 두 가지이다.

- 위치 인수: 작성한 순서대로 매개변수에 전달
- 키워드 인수: `key=value` 형태로 같은 이름의 매개변수에 전달

두 방식을 섞을 때는 위치 인수를 먼저 쓰고, 그 뒤에 키워드 인수를 써야 한다.

```python
def test(a, b):             # argument의 순서대로 parameter에 할당됨.
    print("a:", a)
    print("b:", b)

print("--- 위치 인수 ---")
test(1, 2)

print("--- 키워드 인수 ---")
test(b=30, a=40)            # 지정한 키워드에 맞게 argument가 parameter에 할당됨. (값 명시)

print("--- 위치 인수 + 키워드 인수 ---")
# test(b=200, 100)  # 키워드 인수 뒤에는 위치 인수 작성 불가 --> 위치 인수 + 키워드 인수 순서
# test(100, a=200)  # test() got multiple values for argument 'a'
```

## 5. 매개변수 기본값

기본값이 있는 매개변수는 호출할 때 생략할 수 있다.

단, 기본값이 있는 매개변수는 기본값이 없는 매개변수보다 오른쪽에 작성해야 한다.

```python
def test2(name, age=30, address="한국"):
    print("name:", name, "age:", age, "address:", address)

test2("홍길동", 20) # 인수 누락 시 기본값 (매개변수 란에 지정된) 적용
test2("신사임당", 40, "서울시 서초구") # 인수 존재시 해당 인수 사용
test2("이순신")

# 잘못된 경우
# def wrong(name, age=30, address):   # 디폴트가 아닌 매개변수(address)가 디폴트 매개변수(age) 뒤에 옵니다
```

## 6. packing: `*args`, `**kwargs`

- `*args`: 여러 위치 인수를 하나의 튜플로 모은다.
- `**kwargs`: 여러 키워드 인수를 하나의 딕셔너리로 모은다.
- 매개변수 작성 순서는 일반 매개변수, `*args`, `**kwargs` 순서를 기본으로 생각하면 좋다.

```python
# *: package와 관련된 함수.

def test3(*ars): # *args: 전달 받은 '인자'를 모두 tuple로 묶음 (이름이 꼭 args가 아니어도 됨)
    print(ars)
    return sum(ars) # sum: list, set, tuple에 저장된 모든 값을 더해서 반환

print(test3(1, 2, 3))
print(test3(1,2,3,4,5,6,7,8,9))

# 기본 인수 + *args
def test4(emoji, *args):
    print(emoji, args)

test4("🍗",1,2,3)    # 출력시 emoji는 agrs에 tuple로 같이 묶이지 않음 --> 매개변수 대상이 다르기 때문
test4(1,2,3,4)       # 출력시 1은 agrs에 tuple로 같이 묶이지 않음 --> 매개변수 대상이 다르기 때문
```

```python
def test5(**kwargs): # ** 2개를 묶어서 딕셔너리 형식으로 하나로 묶는다.
    print(kwargs, type(kwargs))     # 매개변수 전달 형태 XXX = XXX

test5()
test5(menu="돈까스", price=10000, options=["치즈","콜라"])


# 기본 인수 + **kwargs
def test6(name, age, **kwargs):
    print("name:", name, "age:", age, "kwargs:", kwargs)

# test6("홍길도", 20, a=2, b=30, 15 = "sdfsd")     ## kwargs는 무조건 일반 인수 뒤에만 작성 가능
# **kwagrs의 입력값은 반드시 변수명 규칙을 준수하는 string(문자열)만 가능
# 숫자로 시작하거나, 예약어 등을 사용한 변수명 사용을 금함 (문자열임에도 불구)


# 일반 인수 + *args + **kwargs
def test7(a,b, *args, **kwargs):
    print("a:", a, "b:", b, "args:", args, "kwargs:", kwargs)

test7(1,2,1,1,1,1,1,1,1,aa=15,bb=85,c=99)
```

## 7. unpacking: `*list`, `**dict`

- `*list`는 리스트의 요소를 꺼내어 위치 인수처럼 나열한다.
- `**dict`는 딕셔너리의 key-value를 꺼내어 키워드 인수처럼 나열한다.

```python
names = ['홍길동', '이순신', '카리나']

def test8(*args):
    print("args:", args)

print("--- list unpacking ---")
test8(names)    # tuple 안에 리스트가 들어있는 형태로 전달됨 -> list로 묶인 상태
test8(*names)   # *: list 앞에 사용되면 unzip이 되고, argument 앞에 사용되면 zip이 됨. 즉, * 하나의 의미는 zip과 unzip 모두 사용 가능함


info = {
    "date" : "2026/06/04",
    "time" : "14:06"
}

def test9(date, time):
    print("date:", date, "time:", time)

test9(info["date"], info["time"]) # 위치 인수
test9(time = info["time"], date = info["date"]) # 키워드 인수
test9(**info)   # **: dict에 대한 packing과 unpacking 모두 지원
#
# print(info)
# print(**info)
```

## 8. 위치 전용과 키워드 전용 매개변수

Python 함수는 특정 매개변수를 위치 전용 또는 키워드 전용으로 제한할 수 있다.

- `/` 앞의 매개변수는 위치 전용이다.
- `*` 뒤의 매개변수는 키워드 전용이다.

```python
def test10(name, / ,greeting="안녕"):
    print(name, greeting)

test10("홍길동")
# test10(name="홍길동", greeting="hello") # TypeError
# -> / 앞 name은 위치 인수로만 동작       <-- 실제로 제한해버리는 매개변수: / ((((위치 인수))))

def test11(a, *, b,c):
    print("a:", a, "b:", b, "c:", c)

# test11(10,20,30) # TypeError
## -> * 뒤는 키워드 인수로만 동작          <-- 실제로 제한해버리는 매개변수: * ((((키워드 인수))))
test11(10, b=20, c=30)
```

## 9. 함수 type hint

함수의 매개변수와 반환값에도 type hint를 작성할 수 있다. type hint는 실행을 강제하기보다 IDE와 정적 분석 도구가 코드를 이해하도록 돕는 역할을 한다.

```python
# 매개 변수 타입 힌트 => a: str, b: int, c: bool
# 반환형 타입 힌트 =>  -> dict[str, object]
def test12(a: str, b:int , c: bool) -> dict:
    return {"a":a, "b":b, "c":c}

print(test12("홍길동", 20, True))
```
