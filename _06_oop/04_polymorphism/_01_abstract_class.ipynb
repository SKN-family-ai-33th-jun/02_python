# 추상클래스 Abstract Class (구체적이지 않은 설계도 == 미완성 설계도)
- 자식클래스에서 구현해야 할 추상메서드를 가진 부모클래스 (인터페이스)
    - 추상 메서드 == 미완성 메서드 == 메서드 정의(<=>선언) 부분이 비어있음

- 추상메소드는 자식클래스에서 반드시 구현해야할 의무를 가진다. (오버라이딩!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!)
- 추상클래스는 직접 객체화 할수 없다. (추상 클래스로 인스턴스 생성 불가)

- abc.ABC클래스를 상속하면 추상클래스를 선언할 수 있다.
    - ABC == Abstract Base Class
- @abstractmethod 데코레이터로 추상메서드를 선언할 수 있다.

```python
from abc import ABC, abstractmethod     # abc: 파이썬 내장모듈, ABC: 추상클래스 만들 때 상속, abstractmethod: 추상 메서드 데코레이터

# 추상 클래스 정의
# - ABC 상속
# == 해당 클래스가 너무 추상적 또는 해당 클래스로 인스턴스 생성 X

# 추상 클래스 목적
# - 상속받은 자식 클래스들에게 공통된 규약(규칙)을 제공하는 용도
class Product(ABC):
    def __init__(self, code: str, name: str, price: int):
        self.__code = code
        self.__name = name
        self.__price = price

    # 추상 메서드 == 미완성 메서드 == 정의 부분이 없음
    @abstractmethod     # 추상 메서드임을 표시
    def get_tax(self):
        pass    # 정의 부분을 상속 받을 자식에게 넘김

    def get_price(self):
        return self.__price

    def __str__(self):
        return f"code={self.__code}, name={self.__name}, price={self.__price}"
```

```python
# 추상 클래스로 인스턴스 생성하기
# prod = Product("111", "노트북", 200000) 추상 클래스는 인스턴스화 할 수 없다.
# TypeError: Can't instantiate abstract class

# 추상 클래스를 상속받은 자식 클래스
class TV(Product):
    def __init__(self, code: str, name: str, price: int, size: int):
        super().__init__(code, name, price)
        self.__size = size

    # 상속 받은 추상 메서드를 재작성 == 오버라이딩
    def get_tax(self):  # 부가세를 10%로 책정해서 반환
        return self.get_price() * 0.1

    # __str__() 오버라이딩
    def __str__(self):
        return f"{super().__str__()}, size={self.__size}"
```

```python
tv1 = TV("samsung-tv-1234", "SAMSUNG OLED TV", 5000000, 85)
print(tv1)
print(tv1.get_tax())
```

## 여러 자식 클래스를 부모 타입으로 다루기

다형성을 이용하면 서로 다른 자식 클래스가 같은 부모 규격을 따르는 모습을 볼 수 있다.

```python
class Book(Product):
    def get_tax(self):
        return 0

    def __str__(self):
        return f"Book({super().__str__()}"

class Food(Product):
    def get_tax(self):
        return self.get_price() * 0.03

    def __str__(self):
        return f"Food({super().__str__()}"


# products 리스트
# - 요소로 Product를 상속 받은 인스턴스만 저장
# - Product가 자식 타입인 TV, Book, Food로 인식되어 Type Hint Error가 발생하지 않는다
#       == Product가 TV, Book, Food 타입으로 변했다 ==     <-- 다형성
products: list[Product] = [
    tv1,
    Book("book-1", "python for beginners", 20000),
    Food("apple-1", "Apple", 10000)
]    # class = 사용자 정의 데이터 타입. 그러므로 Product도 하나의 타입임.

for prod in products:
    print("상품:", prod)
    print("세금:", prod.get_tax())
```
