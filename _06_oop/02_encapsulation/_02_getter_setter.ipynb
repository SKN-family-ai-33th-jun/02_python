## Getter/Setter 메서드
캡슐화된 클래스에서 속성을 제어하기 위한 메서드로써 기본적으로 다음 두 유형이 있다.
1. getter 속성의 값을 읽어내기 위한 메서드 (인스턴스 private 속성의 값을 얻어가게 함)
2. setter 속성의 값을 설정하기 위한 메서드 (인스턴스 private 속성에 값을 대입(세팅))

```python
class MyClass:
    def __init__(self, value):      # 인스턴스 생성시 속성값 초기화 (1회)
        self.__value = value

    # 외부에서 private 속성 값을 얻어가게 하는 메서드  --> 가져가기는 하지만 수정할 수는 없다.
    def get_value(self):
        return self.__value

    # 외부에서 private 속성 값을 변경하는 메서드  --> 값 수정
    def set_value(self, value):     # 인스턴스 생성 후 용도에 따라 값 변경 (다회)
        self.__value = value

my_cls = MyClass(100)
print(my_cls.get_value())

# value값 200으로 세팅
my_cls.set_value(200)
print(my_cls.get_value())
```

## Python의 `@property`

Python에서는 Java식 `get_value()`, `set_value()` 메서드도 사용할 수 있지만, 더 Python다운 방식으로는 `@property`를 자주 사용한다.

`@property`를 사용하면 메서드로 값을 검증하면서도, 사용하는 쪽에서는 속성처럼 읽고 쓸 수 있다.


```python
# @: annotation(어노테이션)      사람이 보는 주석과 같이 annotation은 파이썬이 보는 주석임
                               # 그렇다고 임의로 자유롭게 작성 x  -> 정해진 구문이 있음. @property 등

class Score:
    def __init__(self, value: int):
        self.__value = value

    # 오버로딩: 한 클래스 내에 같은 이름으로 메서드가 여러 개 존재하는 것
    @property
    def value(self):    # get_value()
        return self.__value

    @value.setter       # set_value()
    def value(self, value):
        if not 0 <= value <= 100:
            raise ValueError("점수는 0~100 사이만 세팅 가능")

        self.__value = value

score = Score(80)
print("score.value:", score.value)
# print("score.value:", score.value(90)) # error
score.value = 100   # 메서드 호출이 아닌 변수에 값 대입 하듯 사용 ////////@property 속성
print("score.value:", score.value)

score.value = 120
print("score.value:", score.value)
```
