# 캡슐화
- 속성/메서드를 하나의 클래스에 함께 선언
- private 속성(변수) 선언해서 외부에서 직접 접근을 막고, 제공된 메소드를 통해서 사용하도록 강제
  - private 변수 선언 방법 : 변수명 앞에 __를 붙인다.
- 캡슐화 원칙을 잘 지켰다면, 주요한 데이터가 외부에 노출되지 않은 은닉화 효과까지 얻게된다.

```python
# class: 객체(instance)를 정의한 문서 (속성, 기능<메서드>)           object: 현실 객체,,, instance: 컴퓨터 객체
#\
# 캡슐화: 변수와 함수를 하나로 묶음 + private 속성을 선언하여 외부 접근을 차단함
class Account:
    def __init__(self, username, balance):
        # __가 붙은 속성 == 생성된 인스턴스만 사용 가능한 private 속성 (해당 인스턴스만!!!)
        self.__username = username
        self.__balance = balance    # private 속성에 대하여 언더바 (_)를 2개 입력함 --> __

    # 외부에서 private 속성을 다룰 수 있는 메서드
    def info(self): # 인스턴스 메서드
        print(f"{self.__username}'s balance is {self.__balance}")

    def deposit(self, money):   # 입금 메서드
        # 오입금 방지(0 이하 입금 막기)
        if money <= 0:
            print("입금액이 유효하지 않습니다.")
        else:
            self.__balance += money
            print("입금 금액:", money, "원")
            print("현재 잔액:", self.__balance, "원")

    def withdraw(self, money):  # 출금 메서드
        # 오출금 방지
        if money <= 0:
            print("출금액이 유효하지 않습니다.")
        elif money >= self.__balance:
            print("잔액이 부족합니다.")
        else:
            self.__balance -= money
            print("출금 금액:", money, "원")
            print("현재 잔액:", self.__balance, "원")
```

```python
acc1 = Account("홍길동", 10000)
# print(acc1.username)
# print(acc1.__balance) # 외부 영역에서 클래스 내부 private 요소에 접근이 불가함.

acc1.info()

# 입금 처리
acc1.deposit(50000)
acc1.deposit(0)

# 출금 처리
acc1.withdraw(10_000)
acc1.withdraw(100000)
acc1.withdraw(0)

acc1.info()


##### 결론: 클래스 내부에서 private 변수를 다룰 수 있는 기능에 대한 예제이며, 캡슐화의 기능을 확인할 수 있음. --> __ private 속성
```

## Python private과 name mangling

Python에서 `__balance`처럼 언더스코어 두 개로 시작하는 속성은 외부에서 그대로 접근하기 어렵게 이름이 바뀐다.
이를 name mangling이라고 한다.

다만 이것은 보안을 위한 완전한 접근 제한이라기보다, 클래스 외부에서 직접 사용하지 말라는 강한 신호에 가깝다.
