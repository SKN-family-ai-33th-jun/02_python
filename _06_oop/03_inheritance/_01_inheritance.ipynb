# 상속 Inheritance
- 부모클래스 속성/메서드를 자식클래스에서 선언없이 물려받아 사용하는 것
- 중복된 속성/메서드를 재작성할 필요가 없다.
- 자식클래스를 부모클래스로 묶어 관리할 수 있다.
- 파이썬 다중상속 지원

```python
class Person:
    def __init__(self, name: str, phone: str) -> None:
        self.__name = name
        self.__phone = phone

    def hello(self):
        print(f'안녕하세요, 저는 {self.__name}입니다. 제 연락처는 {self.__phone}이에요.')

    def get_name(self):
        return self.__name

    def get_phone(self):
        return self.__phone

    # __str__(): print(객체)를 했을 때 어떤 문자열을 보여줄지 정하는 메서드
    def __str__(self):
        return f'Person(name={self.__name}, phone={self.__phone})'

p1 = Person("홍길동", "01234567896541230")
print(p1)   # p1에 저장된 주소 출력 ////// but __str__ 활성화시 지정된 데이터 출력됨
```

```python
# Person 상속 받은 자식 클래스 Employee 생성

class Employee(Person):
    def __init__(self, name: str, phone: str, company: str) -> None:
        #부모 생성자 사용: super()
        super().__init__(name, phone)
        self.__company = company

    # method overriding (덮어쓰기)
    # 부모로부터 상속받은 hello 메서드를 재작성 (재정의)
    def hello(self):
        super().hello() # 부모 메서드 호출
        print(f"{self.__company}에서 일하고 있습니다.")  # 부모 메서드 내용 + 추가 내용 작성

    # 오버라이딩 요령: 부모의 메서드를 그대로 재작성하기
    def __str__(self):
        return f'Person(name={self.get_name()}, phone={self.get_phone()}, company={self.__company})'



emp1 = Employee("홍길동", "01025874156", "SKN")
print(emp1.get_phone()) # 상속 받은 메서드 호출
print(emp1.get_name()) # 상속 받은 메서드 호출
emp1.hello()
print(emp1)
```

```python
class Developer(Person):
    # *args == *langs == 3번째 이후 매개변수를 모두 모아 tuple로 저장
    def __init__(self, name: str, phone: str, *langs: tuple[str]) -> None:
        super().__init__(name, phone)
        self.__langs = list(langs)  # tuple -> list 변환

    # langs에 언어 추가
    # - 언어(lang)을 전달 받아와서 langs에 없으면 추가
    # - langs에 이미 존재하는 언어이면 "이미 추가되어 있습니다" 출력
    def add_lang(self, lang: str) -> None:
        if lang not in self.__langs:
            self.__langs.append(lang)
        else:
            print("이미 추가되어 있습니다.")

    # langs에 언어 제거
    # - 언어(lang)을 전달 받아와서 langs에 있으면 제거
    # - langs에 이미 존재하지 않는 언어이면 "존재하지 않는 언어 입니다" 출력
    # - hint: list.remove(요소)
    def remove_lang(self, lang: str) -> None:
        if lang in self.__langs:
            self.__langs.remove(lang)
        else:
            print("존재하지 않는 언어입니다.")

    # hello() 오버라이딩
    # - 먼저 부모의 hello() 출력
    # - "사용 가능한 언어 목록: {langs}" 출력
    def hello(self):
        super().hello()
        print(f"사용가능한 언어는 {self.__langs}입니다.")

    # __str__() 오버라이딩
    # - "Developer(name={name}, phone={phone}, langs={langs})" 출력
    def __str__(self):
        return f"Developer(name={self.get_name()}, phone={self.get_phone()}, langs={self.__langs})"

aa = Developer("홍길동", "01025452565", "python", "JS")
print(aa)
aa.hello()
aa.add_lang("C")
aa.hello()
aa.remove_lang("C++")
aa.hello()
```

## isinstance(), issubclass(), MRO

상속 관계에서는 객체가 어떤 클래스의 인스턴스인지, 클래스가 어떤 부모를 상속하는지 확인할 수 있다.

- `isinstance(객체, 클래스)`: 객체가 해당 클래스 또는 하위 클래스의 인스턴스인지 확인
- `issubclass(자식클래스, 부모클래스)`: 클래스 사이의 상속 관계 확인
- MRO(Method Resolution Order): 메서드를 찾는 순서


```python
# 인스턴스와 출처 클래스 관계 검증
print("isinstance(aa, Developer):", isinstance(aa, Developer)) # aa라는 인스턴스가 Developer 클래스에서 나온 것이 맞는지를 검증함.
print("isinstance(aa, Person):", isinstance(aa, Person)) # aa라는 인스턴스가 Person 클래스에서 나온 것이 맞는지를 검증함. --> 일부라도 있는지를 검증
print("isinstance(aa, Employee):", isinstance(aa, Employee)) # aa라는 인스턴스가 Employee 클래스에서 나온 것이 맞는지를 검증함. --> dev는 Employee에서 나온 것이 아님
print("-" *40)

# 자식클래스와 부모클래스 관계 검증
print("issubclass(Developer, Person):", issubclass(Developer, Person))
print("issubclass(Employee, Person):", issubclass(Employee, Person))
print("issubclass(Employee, Developer):", issubclass(Employee, Developer)) # 형제간의 관계는 남임.

# MRO
print("-" *40)
print("MRO(Developer):", Developer.mro())

# object 클래스 == 모든 종류의 클래스의 부모인 클래스
# class Test: --> 이는 class Test(object):와 동일한 표현임. object를 별도로 기입하지 않으면 default로 자동 추가됨.
```

## Python의 오버라이딩과 오버로딩 주의

Python은 자식 클래스에서 부모 클래스의 메서드를 다시 정의하는 오버라이딩은 자연스럽게 지원한다.
다만 Java처럼 같은 이름의 메서드를 매개변수 조합별로 여러 개 선언하는 메서드 오버로딩은 지원하지 않는다.

같은 클래스 안에서 같은 이름의 메서드를 다시 선언하면, 뒤에 선언한 메서드가 앞의 메서드를 덮어쓴다.


```python
class Calculator:
    def add(self, a):           # 같은 이름 메서드
        return a

    def add(self, a, b):        # 같은 이름 메서드 --> 같은 이름 메서드는 마지막 위치에 있는 메서드만 살아남음
        return a + b


calculator = Calculator()

```
