## 파일 쓰기

```python
# open("파일 경로", 모드(r, w ,a), **options) **: 키워드 매개변수를 dict로 변환
## - 지정된 경로의 파일을 여는 파이썬 내장 함수
## - 열린 파일에 지정된 모드 동작 수행을 할 수 있다.
f = open("./helloworld.txt", "r", encoding="utf-8")
text = f.read()
print(text)
f.close()
```

```python
f = open("helloworld2.txt", "w", encoding="utf-8")
print(f, type(f))

text = """
    옛날 옛날에, 할아버지 할머니가 살았어요. 호랑이는 할아버지의 애완묘였고 늑대는 할머니의 애완견이었어요. 하하하.
"""
f.write(text)   # 지정된 파일에 text 출력
# 기존 내용이 있으면 덮어쓰기
# 지정된 파일이 존재하지 않으면 파일 생성

# 파일 연결에 사용된 메모리 반환
f.close()
```

```python
# with 절
# - with 옆에 작성된 open() 구문에서 사용한 파일 연결을
  # with 절 종료시 자동으로 close()

with open("helloworld2.txt", "w", encoding="utf-8") as f:
    f.write("with 절을 이용하면 자동 close() 된다")
```

## 파일 읽기

```python
with open("helloworld.txt", "r", encoding="utf-8") as f:
    print(f.read()) # 전부 읽어와서 출력
```

```python
# 한 줄씩 읽기
with open("helloworld.txt", "r", encoding="utf-8") as f:
    for i, line in enumerate(f):    # enumerate: 각 줄마다 인덱스를 부여해서 관리할 수 있음
        print(i+1,": ", line, end="") # end="" :  print(). 완료 후 마지막 문자 빈칸
```

```python
# 파일 이어쓰기("a", append 모드)
with open("helloworld.txt", "a", encoding="utf-8") as f:
    f.write("""\n간장공장공장장은 간공장공장장이고 된장공장공장장은 된공장공장장이다""")

with open("helloworld.txt", "r", encoding="utf-8") as f:
    print(f.read())
```

```python
# 파일 복사 == read and write
# 1. 파일 읽기
# 2. 파일 쓰기

with open("helloworld.txt", "r", encoding="utf-8") as f:        # with문은 if나 for문 처럼 지역 변수로 구분되지 않음
    text = f.read() # with 절에 선언한 변수는 지역변수 X --> 전역 사용 가능

with open("helloworld_copy.txt", "w", encoding="utf-8") as f:
    f.write(text)
    print("파일 복사 완료")
```

```python
# 이미지 파일 복사 (아무 파일이나 가능)
# r,w,a: 문자 읽기/문자 쓰기, 문자 이어쓰기
# rb,wb: byte 단위 읽기/byte 단위 쓰기
# encoding: 문자 인코딩

with open("100세 스폰지밥.png", "rb") as f:
    # print(f.read())
    image_bytes = f.read()                  # 사용자 입장에서의 ctrl + c

with open("copy_image.png", "wb") as f:
    f.write(image_bytes)                    # 사용자 입장에서의 ctrl + v
```

## dict-csv 읽고 쓰기

```python
# csv: Comma Seperator Value (comma로 구분된 text 파일임)

# csv 일기
# csv.DictReader(f)     Key:Value
# csv.DictWriter(f, fieldnames)
import csv

with open("users.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)  # csv 모듈 내의 DictReader 클래스에 대한 객체 생성

    # users = []
    # for row in reader:
    #     # print(row, type(row))
    #     users.append(row)

    users = [row for row in reader]

    print(users)
    print(type(users))
```

```python
# csv 쓰기
with open("users.csv", "w", encoding="utf-8", newline="") as f:

    # csv 쓰기 객체 생성
    # - f: 쓸 파일
    # - fieldnames: 첫 줄에 작성될 헤더 (컬럼명)
    writer = csv.DictWriter(f, fieldnames=["id", "name", "email"])

    # users에 사용자 추가
    users.append({"id": 400, "name": "이순신", "email": "sslee@gmail.com"})
    users.append({"id": 500, "name": "강감찬", "email": "kkchan@gmail.com"})

    # 여러 줄 쓰기
    # writer.writerows(users)

    # 한 줄씩 쓰기
    for user in users:
        writer.writerow(user)
```
